```python
import numpy as np

def sample_linear_itf(n_samples=1000):
    # 1. Generate uniform random probabilities (u) between 0 and 1
    u = np.random.uniform(0, 1, n_samples)
    
    # 2. Map through our derived inverse CDF: x = sqrt(u)
    samples = np.sqrt(u)
    
    return samples

# Example usage:
# Generate 5 sampled values
my_samples = sample_linear_itf(n_samples=5000)

# print(my_samples)

import matplotlib.pyplot as plt
plt.hist(my_samples, bins=30, density=True, alpha=0.6, color='g', label='Sampled Data')
plt.show()
```

```python
def sample_linear_itf(n_samples=1000):
    # 1. Generate uniform random probabilities (u) between 0 and 1
    u = np.random.uniform(0, 1, n_samples)
    
    # 2. Map through our derived inverse CDF: x = sqrt(u)
     # Transform uniform samples to follow the desired distribution
    samples = 1./np.sqrt(1-u)  
    
    return samples

samples = sample_linear_itf(n_samples=500000)
plt.hist(samples, bins=30, density=True, alpha=0.6, color='g', label='Sampled Data')
plt.title('Histogram of Samples from Linear ITF')
plt.xlabel('Sample Value')
plt.ylabel('Density')
plt.legend()
plt.yscale('log')
plt.show()  
```
